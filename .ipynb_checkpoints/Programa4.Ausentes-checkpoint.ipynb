**Imputación Valores ausentes:** 
La imputación de valores ausentes consiste en estimar y sustituir datos faltantes con valores plausibles derivados de la información disponible, con el fin de mantener la integridad y validez del análisis

**Formula:**


## 1. Importación de herramientas

Importamos NumPy como `np` para representar los valores faltantes con
`np.nan`. También importamos `SimpleImputer` de scikit-learn para
reemplazar esos valores por la media de su columna.

```python
import numpy as np
from sklearn.impute import SimpleImputer
```

## 2. Preparación de los datos de entrenamiento

Creamos `X_train` con tres filas y dos columnas. En la segunda fila
falta el valor de la primera columna, por lo que lo representamos
con `np.nan`.

```python
# Carga de datos
X_train = [[1,
4],
[np.nan, 3],
[7,
8]]
```

## 3. Imputación con la media de cada columna

Configuramos `SimpleImputer(strategy="mean")` y aplicamos
`fit_transform(X_train)`. Este método calcula la media de cada columna
ignorando los valores faltantes y usa esas medias para rellenarlos.

En la primera columna, la media es `(1 + 7) / 2 = 4`, así que el valor
faltante se reemplaza por `4`. En la segunda columna, la media es
`(4 + 3 + 8) / 3 = 5`. El resultado se guarda en `X_train_clean`.

```python
# Limpieza de datos: imputación valores ausentes (modelo)
imputer = SimpleImputer(strategy='mean') # Utilizamos la estrategia de la media
# Aplicar los métodos "fit" y "transform" para imputar los valores pérdidos de␣
#X_train
X_train_clean = imputer.fit_transform(X_train)
print(X_train_clean)
```

## 4. Preparación de los datos de prueba

Creamos `X_test` con tres filas y las mismas dos columnas, en el mismo
orden que en `X_train`. Hay dos valores faltantes: uno en la primera
columna de la primera fila y otro en la segunda columna de la segunda fila.

```python
# Imputación de valores ausentes sobre el conjunto de test
X_test = [[np.nan, 10],
[6,
np.nan],
[8,
2]]
```

## 5. Aplicación de la imputación a los datos de prueba

Usamos `imputer.transform(X_test)` para rellenar los valores faltantes
con las medias aprendidas en el entrenamiento. El primer valor faltante
se reemplaza por `4` y el segundo por `5`.

Guardamos el resultado en `X_test_clean`. Así, los datos de prueba se
preparan con los mismos parámetros que los datos de entrenamiento.

```python
# Imputar los datos del conjunto de test
X_test_clean=imputer.transform(X_test)
print(X_test_clean)
```
