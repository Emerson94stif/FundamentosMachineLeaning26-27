**Hold Out:** El método Hold Out es una técnica de validación que divide el dataset en conjuntos de entrenamiento y de prueba (o retención) para evaluar de forma imparcial el rendimiento de un modelo en datos no vistos

**Formula:**

## 1. Importación de herramientas

Importamos NumPy para contar los ejemplos de cada clase. De scikit-learn
importamos `datasets` para cargar Iris, `train_test_split` para dividir
los datos, `StandardScaler` para estandarizar las variables y
`DummyClassifier` para construir un modelo de referencia.

```python
import numpy as np
from sklearn import datasets
from sklearn.dummy import DummyClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

## 2. Carga del conjunto de datos Iris

Cargamos Iris con `datasets.load_iris()` y mostramos los nombres de las
tres especies que queremos predecir: `setosa`, `versicolor` y `virginica`.

```python
# Carga de datos
iris = datasets.load_iris()
print(iris.target_names)
```

## 3. Características y distribución de las clases

Consultamos `iris.data.shape` y encontramos 150 flores y cuatro variables
de entrada. `iris.target` contiene las etiquetas `0`, `1` y `2`,
que representan las tres especies.

Con `np.unique(..., return_counts=True)` contamos cuántas flores
pertenecen a cada especie. El conjunto original está equilibrado,
con 50 ejemplos por clase.

```python
# Mostrar características de la tabla de datos.
print("Tabla de datos: %d instancias y %d atributos" % (iris.data.shape[0], iris.data.shape[1]))
print("Valores de la clase:", set(iris.target))

# Cuantificamos el número de instancias que contiene el dataset por clase
valores, ocurrencias = np.unique(iris.target, return_counts=True)
print(valores, ocurrencias)
```

## 4. Separación del conjunto de prueba

Dividimos las 150 flores en 120 registros para desarrollar el modelo
y 30 para la prueba final. Después contamos las flores de cada clase
en el conjunto de prueba.

```python
# Test: hold-out split 80-20%. # Partición externa
X_training, X_test, y_training, y_test = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
valores_test, ocur_test = np.unique(y_test, return_counts=True)
print('Test: ', 'clases:', valores_test, ' ocurrencias: ', ocur_test)
```

## 5. Estandarización de las variables

Usamos `StandardScaler` para restar la media y dividir por la desviación
estándar de cada variable. En esta celda, `fit_transform()` calcula
esos parámetros con `X_training` y transforma sus datos. Luego,
`transform()` aplica los mismos parámetros a `X_test`.

```python
# Estandarizar las características de entrenamiento y de test
standardizer = StandardScaler()
X_training = standardizer.fit_transform(X_training)
X_test = standardizer.transform(X_test)
```

## 6. Separación del conjunto de validación

Dividimos los 120 registros de `X_training` en 96 para entrenar
(`X_train`) y 24 para validar (`X_val`). También mostramos cuántas
flores de cada especie quedaron en ambos grupos.

```python
# Validación: hold-out split 80-20%. # Partición interna
X_train, X_val, y_train, y_val = train_test_split(X_training, y_training, test_size=0.2, random_state=42)
valores_train, ocur_train = np.unique(y_train, return_counts=True)
print('Entrenamiento: ', ' clases:', valores_train, '  ocurrencias:', ocur_train)

valores_val, ocur_val = np.unique(y_val, return_counts=True)
print('Validation:    ', ' clases:', valores_val, '  ocurrencias:', ocur_val)
```

## 7. Creación del modelo de referencia

Configuramos `DummyClassifier` con la estrategia `prior`. Al predecir
una clase, este clasificador elige siempre la más frecuente del
entrenamiento. Su exactitud sirve como referencia para evaluar el SVC.

```python
# Construcción del objeto que contiene el algoritmo de aprendizaje.
clf = DummyClassifier(strategy='prior', random_state=42)
```

## 8. Entrenamiento del modelo de referencia

Ejecutamos `clf.fit(X_train, y_train)` para que el clasificador
identifique la clase más frecuente. En esta partición fue la clase `2`
(`virginica`), con 34 ejemplos de los 96 de entrenamiento.

```python
# Entrenamiento del algoritmo de aprendizaje.
clf = clf.fit(X_train, y_train)
```

## 9. Evaluación del modelo de referencia

Usamos `score()` para medir la proporción de predicciones correctas.
El modelo obtuvo una exactitud del 20,83 % en validación y del 36,67 %
en prueba. Como siempre predice `virginica`, su exactitud en cada
conjunto coincide con la proporción de flores de esa especie.

```python
# Evaluación del algoritmo de aprendizaje con el método "score" que devuelve directamente la métrica de 'accuracy'
val_accuracy = clf.score(X_val, y_val)
print("Exactitud en validación: ", val_accuracy)

test_accuracy = clf.score(X_test, y_test)
print("Exactitud en test: ", test_accuracy)
```

## 10. Comparación de las predicciones con las etiquetas reales

Usamos `predict()` para generar las predicciones de validación y de
prueba. Comparamos `y_pred_val` con `y_val` en validación, y `y_pred_test`
con `y_test` en prueba.

```python
# Obtenemos las predicciones sobre conjunto de validación y de test
y_pred_val = clf.predict(X_val)
print('Predicciones de validación ', y_pred_val)
print('Etiquetas reales validación', y_val)

y_pred_test = clf.predict(X_test)
print('\nPredicciones de test ', y_pred_val)
print('Etiquetas reales test', y_val)
```

## 11. Entrenamiento y evaluación del clasificador SVC

Creamos un clasificador `SVC` con `C=0.5` y lo entrenamos con
`X_train` y `y_train`. El modelo aprende a distinguir las especies
a partir de las cuatro medidas de cada flor.

En esta ejecución, el SVC obtuvo una exactitud del 91,67 % en
validación y del 96,67 % en prueba. Superó al modelo de referencia
en los dos conjuntos.

```python
# Aplicamos un ejemplo con un clasificador más complejo que el "dummyclassifier"
from sklearn.svm import SVC
svc = SVC(C=0.5) # Definimos algoritmo
svc.fit(X_train, y_train) # Entrenamos modelo

val_accuracy = svc.score(X_val, y_val) # Evaluamos modelo en validación
print('Exactitud en validación ', np.round(val_accuracy*100, 4), '%')

test_accuracy = svc.score(X_test, y_test) # Evaluamos modelo en test
print('Exactitud en test ', np.round(test_accuracy*100, 4), '%')
```

### ¿Qué hace este código?

Este bloque de código se encarga de evaluar un modelo de clasificación mediante la obtención de predicciones sobre los conjuntos de validación y de test[cite: 10]. 

A continuación se detalla su funcionamiento paso a paso:

* **Predicción en validación (`X_val`):**
  * Utiliza el método `.predict()` del clasificador `svc` para obtener las predicciones del conjunto de validación (`y_pred_val`)[cite: 10].
  * Imprime en consola el texto `'Predicciones de validación '` junto con el arreglo `y_pred_val`[cite: 10].
  * Imprime en consola el texto `'Etiquetas reales validación'` junto con las etiquetas reales `y_val`[cite: 10].

* **Predicción en prueba (`X_test`):**
  * Utiliza el método `.predict()` del clasificador `svc` para obtener las predicciones del conjunto de test (`y_pred_test`)[cite: 10].
  * Imprime en consola el texto `'\nPredicciones de test '` junto con el arreglo `y_pred_test`[cite: 10].
  * Imprime en consola el texto `'Etiquetas reales test'` junto con las etiquetas reales `y_test`[cite: 10].

```python
# Obtenemos Las predicciones sobre conjunto de validación y de test
y_pred_val = svc.predict(X_val)
print('Predicciones de validación ', y_pred_val)
print('Etiquetas reales validación', y_val)

y_pred_test = svc.predict(X_test)
print('\nPredicciones de test ', y_pred_test)
print('Etiquetas reales test', y_test)
```

## 12. Guardado y carga del modelo

Guardamos el clasificador entrenado en `models/svc.pickle` con
`pickle.dump()` y lo recuperamos con `pickle.load()`. Para reutilizarlo,
asignamos el resultado de la carga a una variable, por ejemplo,
`svc_cargado = pickle.load(fr)`.

Si queremos predecir con medidas nuevas, necesitamos conservar también
el `StandardScaler` ajustado para aplicar la misma transformación.

```python
# Guardar modelo
import pickle
from pathlib import Path

Path("models").mkdir(exist_ok=True)
with open('models/svc.pickle', 'wb') as fw:
    pickle.dump({
        "modelo": svc,
        "estandarizacion": standardizer,
        "clases": iris.target_names
    }, fw)
```

```python
#Cargamos el modelo guardado
with open("models/svc.pickle", "rb") as archivo:
    datos = pickle.load(archivo)

#Cargamos cada uno de los atributos
modelo = datos["modelo"]
estandarizador = datos["estandarizacion"]
clases = datos["clases"]
```

### ¿Qué hace este código?

Este bloque de código se encarga de capturar nuevos datos introducidos por el usuario para prepararlos y utilizarlos posteriormente en un modelo predictivo basado en el conjunto de datos Iris.

A continuación se detalla su funcionamiento paso a paso:

* **Ingreso de datos por teclado:**
  * Solicita al usuario mediante funciones `input` que introduzca cuatro medidas físicas de la flor: el largo del sépalo, el ancho del sépalo, el largo del pétalo y el ancho del pétalo.
  * Convierte cada valor ingresado de texto a un número decimal utilizando la función `float`.

* **Estructuración en arreglo de NumPy:**
  * Organiza las variables recolectadas dentro de un arreglo bidimensional con `np.array`.
  * Mantiene estrictamente el mismo orden de características de entrada que utiliza el dataset Iris original.

* **Impresión en pantalla:**
  * Muestra un mensaje en la consola con el arreglo resultante (`entrada`) para que el usuario pueda verificar los valores que acaba de digitar.

```python
# Ingresamos los datos por teclado
largo_sepalo = float(input("Ingrese el largo del sepalo (cm): "))
ancho_sepalo = float(input("Ingrese el ancho del sepalo (cm): "))
largo_petalo = float(input("Ingrese el largo del petalo (cm): "))
ancho_petalo = float(input("Ingrese el ancho del petalo (cm): "))

# Mantener el mismo orden de características de entrada del data set Iris
entrada = np.array([[
    largo_sepalo,
    ancho_sepalo,
    largo_petalo,
    ancho_petalo
]])

# Imprimimos los datos de entrada
print("Los datos de entrada son: ", entrada)
#transformamos los datos de netrada con la operación de estandarización
entrada_estadarizada=estandarizador.transform(entrada)
```

```python
print(entrada_estadarizada)
```

```python
# generamos la nueva predicción

prediccion= modelo.predict(entrada_estadarizada)
clase = int (prediccion[0])
print("clase predicha:", clase)
print("especie predicha:", clases[clase])
```
