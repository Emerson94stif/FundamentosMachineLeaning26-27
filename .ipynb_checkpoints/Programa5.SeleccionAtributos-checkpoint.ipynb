**Seleccion Atributos:** 
La selección de atributos es el proceso de elegir un subconjunto de las variables originales que maximize el rendimiento del modelo y minimicen redundancia o irrelevancia

**Formula:**

## 1. Importación de herramientas

Importamos NumPy como `np` y cargamos `VarianceThreshold` desde
`sklearn.feature_selection`. Usaremos `VarianceThreshold` para
seleccionar columnas según cuánto varían sus valores.

```python
import numpy as np
from sklearn.feature_selection import VarianceThreshold
```

## 2. Creación de la matriz de datos

Definimos `X` con seis filas y tres columnas. Cada fila representa
una observación y cada columna representa un atributo binario,
con valores `0` y `1`. Mostramos la matriz original con `print(X)`.

```python
# Carga de datos.
X = [[0, 0, 1], [0, 1, 0], [1, 0, 0],
[0, 1, 1], [0, 1, 0], [0, 1, 1]]
print(X)
```

## 3. Selección de atributos por varianza

Definimos `th = 0.2` como umbral de varianza. Es un valor de varianza
en la escala de los datos. El método `fit_transform(X)` calcula
la varianza de cada columna y conserva las que superan el umbral.
La matriz resultante tiene las mismas seis filas y dos columnas.
Este método es no supervisado porque selecciona los atributos
sin usar una variable objetivo.

```python
# Limpieza de datos: selección de atributos.
# fijamos un umbral (th) para decidir qué porcentaje mínimo deben variar los␣
#valores de una columna para considerarse
# una columna relevante a la hora de mantenerla en nuestro estudio
th = 0.2
sel = VarianceThreshold(th)
print(sel.fit_transform(X))
```
