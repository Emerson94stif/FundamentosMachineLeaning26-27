**Éxito verificable.** El análisis debe conservar el conjunto de entrada con más de 100 000 registros, documentar las 50 columnas, faltantes, duplicados e IQR, producir los gráficos pedidos, comprobar numéricamente ambas transformaciones y exportar CSV con índice de referencia.

## 3. Comprensión de los datos

Se verifica la diferencia entre hospitalizaciones y pacientes y se construye el diccionario completo con la descripción de UCI, el tipo observado y el significado conceptual.

```python
numeros = [
    "time_in_hospital", "num_lab_procedures", "num_procedures",
    "num_medications", "number_outpatient", "number_emergency",
    "number_inpatient"
]
print("Tamaño de los datos:", df.shape)
print("Hospitalizaciones:", len(df))
print("Pacientes únicos:", df.patient_nbr.nunique())

dic = dataset.variables[["name", "description", "type", "role"]].copy()
dic["tipo_observado"] = dic["name"].map(df.dtypes.astype(str))
dic["tipo_conceptual"] = np.select(
    [dic.role.eq("ID"), dic.type.eq("Integer")],
    ["identificador", "cuantitativa"], default="categórica"
)
dic["unidades_o_categorías"] = [
    "días" if nombre == "time_in_hospital" else
    "intervalos de 10 años" if nombre == "age" else
    "NO, <30, >30" if nombre == "readmitted" else
    "código de referencia" if nombre in {"patient_nbr", "encounter_id"} else
    "conteo" if tipo == "cuantitativa" else
    "categorías o códigos según la descripción"
    for nombre, tipo in zip(dic.name, dic.tipo_conceptual)
]
dic["rol_analítico"] = np.select(
    [dic.role.eq("ID"), dic.role.eq("Target")],
    ["identificación", "referencia de readmisión"],
    default="característica"
)
dic = dic.rename(columns={"name": "nombre", "description": "significado"})
display(dic[["nombre", "significado", "tipo_observado", "tipo_conceptual", "unidades_o_categorías", "rol_analítico"]])
assert len(dic) == df.shape[1] == 50
```

Se cargaron **101 766 hospitalizaciones**, **50 columnas** y **71 518 pacientes identificados de forma distinta**. La unidad de análisis es la hospitalización. Las columnas de tipo entero que representan identificadores o categorías no son mediciones continuas; `age` también se conserva como intervalos.

```python
reporte_ausencias = pd.concat([
    df.isna().sum().rename("nulos"),
    df.eq("?").sum().rename("marcadores_?")
], axis=1)
reporte_ausencias["ausentes_observados"] = reporte_ausencias.sum(axis=1)
reporte_ausencias["porcentaje"] = 100 * reporte_ausencias.ausentes_observados / len(df)
reporte_ausencias.index.name = "columna"
display(reporte_ausencias.reset_index().round(2))

print("Filas completas repetidas:", df.duplicated().sum())
print("Encuentros repetidos:", df.encounter_id.duplicated().sum())
print("Registros posteriores del mismo paciente:", df.patient_nbr.duplicated().sum())
```

Los faltantes verdaderos aparecen en `race` (2 273), `weight` (98 569), `payer_code` (40 256), `medical_specialty` (49 949) y los tres diagnósticos (21, 358 y 1 423). Los porcentajes más altos corresponden a `weight` (96,86 %), `medical_specialty` (49,08 %) y `payer_code` (39,56 %). La carga no muestra `?` literales.

La fuente describe `None` en `max_glu_serum` y `A1Cresult` como **prueba no realizada**. La lectura lo presenta como nulo, por lo que se recuperará como categoría válida en una copia. No hay filas idénticas ni identificadores de encuentro repetidos. Hay 30 248 registros con `patient_nbr` ya visto; pueden representar hospitalizaciones diferentes y se conservan.

```python
resumen_numerico = df[numeros].describe(percentiles=[.25, .5, .75]).T
resumen_numerico = resumen_numerico.rename(columns={
    "25%": "Q1", "50%": "mediana", "75%": "Q3", "std": "desviación"
})[["min", "Q1", "mediana", "mean", "Q3", "max", "desviación"]]
display(resumen_numerico.round(2))

clases = df.readmitted.value_counts().reindex(["NO", ">30", "<30"])
display(pd.DataFrame({
    "frecuencia": clases,
    "porcentaje": 100 * clases / len(df)
}).round(2))
```

`num_procedures` tiene mediana **1**, media **1,34**, Q1 **0**, Q3 **2** y máximo **6**. `number_inpatient` tiene mediana **0**, media **0,64**, Q1 **0**, Q3 **1** y máximo **21**; muchas hospitalizaciones presentan cero ingresos previos. La categoría `NO` representa **54 864** encuentros (53,91 %), `>30` **35 545** (34,93 %) y `<30` **11 357** (11,16 %).

```python
pareja = ["num_procedures", "number_inpatient"]
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for fila, variable in enumerate(pareja):
    sns.histplot(data=df, x=variable, bins=20, discrete=True, color="#59478f", ax=axes[fila, 0])
    axes[fila, 0].set(title=f"Frecuencias de {variable}", ylabel="Hospitalizaciones")
    sns.boxplot(data=df, x=variable, color="#c2b6e3", ax=axes[fila, 1])
    axes[fila, 1].set(title=f"Dispersión de {variable}")
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(7, 4))
ax.bar(clases.index, clases.values, color=["#59478f", "#8974bd", "#b1a3d4"])
ax.set(xlabel="Categoría", ylabel="Hospitalizaciones", title="Frecuencia de readmisión")
plt.tight_layout()
plt.show()
```

**Lectura de los gráficos.** En el histograma de `num_procedures` predominan las cantidades bajas, y su caja muestra una mediana de 1 y una mitad central entre 0 y 2; los valores de 6 aparecen más allá del umbral usual. El histograma de `number_inpatient` concentra muchos encuentros en 0, con una cola hacia valores altos; su caja también representa una mediana de 0, con mitad central de 0 a 1. El gráfico de barras ordena las categorías como `NO`, `>30` y `<30` de mayor a menor frecuencia. Un punto atípico no implica por sí mismo un registro incorrecto.

```python
cuartil_1 = df[pareja].quantile(.25)
cuartil_3 = df[pareja].quantile(.75)
distancia_iqr = cuartil_3 - cuartil_1
limite_bajo = cuartil_1 - 1.5 * distancia_iqr
limite_alto = cuartil_3 + 1.5 * distancia_iqr
fuera_del_rango = df[pareja].lt(limite_bajo) | df[pareja].gt(limite_alto)

tabla_iqr = pd.DataFrame({
    "Q1": cuartil_1, "Q3": cuartil_3, "IQR": distancia_iqr,
    "límite_inferior": limite_bajo, "límite_superior": limite_alto,
    "cantidad_atípica": fuera_del_rango.sum(),
    "porcentaje_atípico": 100 * fuera_del_rango.mean()
})
display(tabla_iqr.round(2))
```

El IQR identifica **4 954 hospitalizaciones (4,87 %)** con 6 procedimientos, superiores al límite 5, y **7 049 (6,93 %)** con al menos 3 hospitalizaciones anteriores, por encima del límite 2,5. Los dos límites inferiores son negativos y no marcan conteos negativos. Se conservan estos registros porque el IQR describe rareza estadística, no verifica si una observación es un error. Los extremos influirán en el escalamiento y por eso permanecen documentados.

```python
desglose = df.groupby("readmitted")[pareja].agg(["mean", "median"])
display(desglose.round(2))
```

**Relación descriptiva con readmisión.** En `<30`, la media de hospitalizaciones previas es **1,22**, frente a **0,84** en `>30` y **0,38** en `NO`. Las tres medianas son 0. Para `num_procedures`, la media es mayor en `NO` (**1,41**) que en `<30` (**1,28**) y `>30` (**1,25**); la mediana es 1 en todos los grupos. Se comparan hospitalizaciones, sin atribuir una causa a estas diferencias.

## 4. Preparación de los datos

La regla de limpieza conserva filas válidas y registra el tamaño antes y después. `None` se recupera como categoría para las pruebas no realizadas. Las columnas ajenas a las siete variables de escalamiento pueden conservar sus faltantes; en estas siete se comprobará si la imputación es necesaria. Los identificadores se mantienen como referencia sin transformarlos.

```python
trabajo = df.copy()
antes = trabajo.shape
antes_pruebas = trabajo[["max_glu_serum", "A1Cresult"]].isna().sum()

trabajo[["max_glu_serum", "A1Cresult"]] = trabajo[["max_glu_serum", "A1Cresult"]].fillna("None")

print("Filas y columnas antes:", antes)
print("Filas y columnas después:", trabajo.shape)
display(pd.DataFrame({
    "nulos_antes": antes_pruebas,
    "nulos_después": trabajo[["max_glu_serum", "A1Cresult"]].isna().sum()
}))
print("Nulos en siete variables:", int(trabajo[numeros].isna().sum().sum()))
print("Conteos negativos:", int(trabajo[numeros].lt(0).sum().sum()))
print("Duración fuera de 1 a 14 días:", int((~trabajo.time_in_hospital.between(1, 14)).sum()))
print("Filas duplicadas:", int(trabajo.duplicated().sum()))
assert trabajo.shape == antes == (101766, 50)
```

La copia mantiene **101 766 filas y 50 columnas**. Se recuperaron 96 420 categorías `None` en `max_glu_serum` y 84 748 en `A1Cresult`; esto registra que la prueba no se realizó y no es una imputación de su resultado. En las siete variables seleccionadas hay **cero nulos y cero negativos**. La duración hospitalaria observada permanece entre 1 y 14 días. Se conservan los atípicos y las hospitalizaciones repetidas de un mismo paciente.

```python
identificadores = trabajo.loc[:, ["encounter_id", "patient_nbr"]].copy()
base = trabajo.loc[:, numeros].copy()
print("Antes de seleccionar:", trabajo.shape)
print("Después de seleccionar:", base.shape)
display(base.head())
assert base.index.equals(identificadores.index)
assert not base.isna().any().any()
```

Se preservó una copia de las siete mediciones antes de transformarlas. `encounter_id` y `patient_nbr` se excluyen por ser identificadores. También se excluyen del escalamiento los códigos categóricos `admission_type_id`, `discharge_disposition_id` y `admission_source_id`, los intervalos de `age` y la referencia `readmitted`.

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler(feature_range=(0, 1))
datos_01 = pd.DataFrame(minmax.fit_transform(base.copy()), columns=base.columns, index=base.index)
display(datos_01.head(10).round(4))

limites = pd.concat([
    base.min().rename("mínimo_antes"), base.max().rename("máximo_antes"),
    datos_01.min().rename("mínimo_después"), datos_01.max().rename("máximo_después")
], axis=1)
display(limites.round(4))
assert datos_01.ge(-1e-9).all().all() and datos_01.le(1 + 1e-9).all().all()
assert datos_01.index.equals(base.index)
```

**Normalización.** `MinMaxScaler` ajusta cada columna al rango [0, 1]. Para una variable no constante aplica `(x − mínimo) / (máximo − mínimo)`. Los extremos conservados fijan ese rango y pueden comprimir gran parte de los datos. Si una columna fuera constante, la expresión manual dividiría para cero; el escalador gestiona ese caso y en [0, 1] produce ceros.

```python
from sklearn.preprocessing import StandardScaler

zscore = StandardScaler()
datos_z = pd.DataFrame(zscore.fit_transform(base.copy()), columns=base.columns, index=base.index)
display(datos_z.head(10).round(4))

revisión_z = pd.DataFrame({"media": datos_z.mean(), "desviación_ddof_0": datos_z.std(ddof=0)})
display(revisión_z.round(4))
variables_constantes = base.nunique().eq(1)
assert datos_z.mean().abs().lt(1e-9).all()
assert (datos_z.loc[:, ~variables_constantes].std(ddof=0) - 1).abs().lt(1e-9).all()
assert datos_z.loc[:, variables_constantes].std(ddof=0).abs().lt(1e-9).all()
assert datos_z.index.equals(base.index)
```

**Estandarización.** `StandardScaler` se ajustó sobre otra copia de los datos limpios, sin usar la versión Min-Max. Para variables no constantes aplica `z = (x − media) / desviación estándar`. Las medias transformadas son cercanas a 0 y las desviaciones poblacionales (`ddof=0`) cercanas a 1. Pueden surgir valores negativos y mayores que 1. Una columna constante produciría ceros y desviación resultante 0.

```python
vistas = [("Original", base), ("Min-Max", datos_01), ("Z-score", datos_z)]
estadisticas_comparadas = []
for etiqueta, marco in vistas:
    tabla = marco.agg(["min", "max", "mean"]).T
    tabla["desviación_ddof_0"] = marco.std(ddof=0)
    tabla.insert(0, "versión", etiqueta)
    tabla.insert(0, "variable", tabla.index)
    estadisticas_comparadas.append(tabla.reset_index(drop=True))
estadisticas_comparadas = pd.concat(estadisticas_comparadas, ignore_index=True)
display(estadisticas_comparadas.round(4))

fig, axes = plt.subplots(2, 3, figsize=(15, 7), sharey="row")
for fila, variable in enumerate(pareja):
    for columna, (etiqueta, marco) in enumerate(vistas):
        axes[fila, columna].hist(marco[variable], bins=20, color="#59478f", edgecolor="white")
        axes[fila, columna].set(title=f"{variable}: {etiqueta}", xlabel="Valor", ylabel="Hospitalizaciones")
plt.tight_layout()
plt.show()
```

La tabla estadística compara mínimos, máximos, medias y desviaciones de las siete variables en las tres versiones. Min-Max cambia cada mínimo a 0 y máximo a 1. Z-score centra cada variable en 0 y le asigna desviación 1. En los gráficos de las dos variables, las barras y sus frecuencias conservan su forma: solo cambia la escala horizontal. El cambio lineal de escala no corrige asimetría ni elimina atípicos.

```python
elegida = "num_medications"
cinco = [1, 8, 15, 20, 30]
minimo = base[elegida].min()
maximo = base[elegida].max()
media = base[elegida].mean()
sigma = base[elegida].std(ddof=0)
print(f"{elegida}: mínimo {minimo}, máximo {maximo}, media {media:.10f}, sigma {sigma:.10f}")

filas = [base.index[base[elegida].eq(x)][0] for x in cinco]
operaciones = pd.DataFrame({
    "índice_original": filas, "x": cinco,
    "Min-Max fórmula": [(x - minimo)/(maximo - minimo) for x in cinco],
    "Min-Max Python": datos_01.loc[filas, elegida].to_numpy(),
    "Z-score fórmula": [(x - media)/sigma for x in cinco],
    "Z-score Python": datos_z.loc[filas, elegida].to_numpy()
})
display(operaciones.round(4))
assert np.allclose(operaciones["Min-Max fórmula"], operaciones["Min-Max Python"])
assert np.allclose(operaciones["Z-score fórmula"], operaciones["Z-score Python"])
```

**Cinco demostraciones con `num_medications`.** Mínimo **1**, máximo **81**, media **16,0218442309**, desviación poblacional **8,1275262764**.

| Medicamentos | Operación Min-Max | Resultado | Operación Z-score | Resultado |
|---:|---|---:|---|---:|
| 1 | (1 − 1) / (81 − 1) | 0,0000 | (1 − 16,0218442309) / 8,1275262764 | −1,8483 |
| 8 | (8 − 1) / (81 − 1) | 0,0875 | (8 − 16,0218442309) / 8,1275262764 | −0,9870 |
| 15 | (15 − 1) / (81 − 1) | 0,1750 | (15 − 16,0218442309) / 8,1275262764 | −0,1257 |
| 20 | (20 − 1) / (81 − 1) | 0,2375 | (20 − 16,0218442309) / 8,1275262764 | 0,4895 |
| 30 | (30 − 1) / (81 − 1) | 0,3625 | (30 − 16,0218442309) / 8,1275262764 | 1,7199 |

Las cinco comparaciones coinciden con la salida de Python al redondear a cuatro decimales.

```python
datos_01.to_csv("datos_normalizados.csv", index=True, index_label="indice_original")
datos_z.to_csv("datos_estandarizados.csv", index=True, index_label="indice_original")

archivo_01 = pd.read_csv("datos_normalizados.csv", index_col="indice_original")
archivo_z = pd.read_csv("datos_estandarizados.csv", index_col="indice_original")
print("Normalizado:", archivo_01.shape)
print("Estandarizado:", archivo_z.shape)
assert archivo_01.shape == archivo_z.shape == (101766, 7)
assert archivo_01.index.equals(archivo_z.index)
assert archivo_01.index.equals(identificadores.index)
assert archivo_01.columns.tolist() == archivo_z.columns.tolist() == numeros
print("Correspondencia de filas y siete columnas verificada.")
```

## 5. Conclusiones y reflexión

Esta exploración verificó **101 766 hospitalizaciones**, **71 518 pacientes distintos** y el significado de las 50 columnas. `NO` concentra el 53,91 % de los encuentros. La media de hospitalizaciones anteriores resulta mayor en `<30` que en `NO`, mientras la media de procedimientos del encuentro es ligeramente mayor en `NO`. No se puede deducir una causa de estas diferencias descriptivas.

El IQR señaló 4 954 valores altos de `num_procedures` y 7 049 de `number_inpatient`, conservados por falta de evidencia de error. Las siete mediciones seleccionadas no requirieron imputación. Los CSV exportados tienen las mismas 101 766 filas y siete variables, con el índice original como referencia.

Min-Max conviene cuando interesa el intervalo [0, 1], mientras Z-score expresa la distancia a la media en desviaciones. Ambas transformaciones conservan la forma de las distribuciones y son sensibles a valores extremos. Los resultados describen el periodo 1999–2008 y los hospitales representados; no son diagnósticos.

**Preguntas de reflexión:** (1) Un código numérico categórico nombra una clase y no mide una magnitud. (2) Un valor extremo determina parte del rango Min-Max y afecta la media y la desviación Z-score. (3) Dos hospitalizaciones del mismo paciente pueden ser encuentros válidos y diferentes. (4) Escalar cambia unidades y ubicación numérica, sin corregir automáticamente la asimetría. (5) Si luego se entrenara un modelo, primero se separarían los datos, se ajustarían los escaladores solo en entrenamiento y se evitaría que un mismo paciente aparezca en diferentes particiones.
