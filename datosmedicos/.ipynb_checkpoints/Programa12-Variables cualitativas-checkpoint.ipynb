```python
#importamos librerias  para la creación de datos
import numpy as np
import random 
import matplotlib.pyplot as plt
import numpy as np
random.seed(42)
```

```python
# creo la estructut¿ra de datos
nivel_estudios = []
nivel_estudios.extend(['Secundaria']*110)
nivel_estudios.extend(['Bachillerato']*172)
nivel_estudios.extend(['Grado']*145)
nivel_estudios.extend(['Master']*52)
nivel_estudios.extend(['Doctorado']*21)
nivel_estudios = np.random.permutation(nivel_estudios)## mezcla todos los datos en desorden
```

```python
print(nivel_estudios)
```

```python
#realizamos diagrama de barras para entender como estan los datos

estudios, frecuencia = np.unique(nivel_estudios, return_counts=True)
print(estudios, frecuencia)
```

```python
plt.ylabel('Frecuencia absoluta')
plt.xlabel('Nivel de estudios')
plt.title('Diagrama de barras de estudio en la cuidad de Valencia')
plt.bar(estudios, frecuencia) # https://matplotlib.org/stable/api/_as_gen/
#matplotlib.pyplot.bar.html
plt.show()
```

```python
# DIAGRAMA DE SECTORES
colores = {'yellow', 'red', 'green', 'blue', 'magenta'}
plt.pie(frecuencia, labels=estudios, colors=colores, autopct='%1.2f%%')
plt.title('Diagrama de sectores')
plt.show()
```
