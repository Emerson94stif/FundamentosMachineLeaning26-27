**Situación.** Un equipo hospitalario necesita describir las características de las hospitalizaciones de personas con diabetes y comprender la distribución de las categorías de readmisión antes de plantear modelos. Los equipos de atención y análisis son los interesados en la calidad y presentación de estos datos.

**Objetivo general.** Revisar la calidad del conjunto y producir dos versiones comparables de siete mediciones cuantitativas: una normalizada y otra estandarizada.

**Preguntas.** ¿Cómo se distribuyen los procedimientos de laboratorio y los medicamentos, incluidos sus valores extremos? ¿Qué diferencias descriptivas se observan entre las categorías de `readmitted`?

**Criterios verificables.** Confirmar más de 100 000 hospitalizaciones; documentar las 50 columnas, ausencias y duplicados; presentar las estadísticas y los gráficos solicitados; comprobar el rango Min-Max y la media y desviación Z-score; exportar dos CSV con el mismo orden de filas.

**Variable de referencia.** `NO` significa que no hubo readmisión, `<30` que ocurrió en menos de 30 días y `>30` que ocurrió después de 30 días. Esta variable no se escala.

## Fase 2. Comprensión de los datos

Primero se diferencia la cantidad de hospitalizaciones de la cantidad de pacientes únicos. Después se integra el diccionario publicado por UCI con los tipos observados tras la carga.

```python
print("Hospitalizaciones:", len(df))
print("Pacientes únicos:", df.patient_nbr.nunique())
print("Columnas:", df.shape[1])

cuantitativas = [
    "time_in_hospital", "num_lab_procedures", "num_procedures",
    "num_medications", "number_outpatient", "number_emergency",
    "number_inpatient"
]

info_uci = dataset.variables.set_index("name").reindex(df.columns)
tipos_conceptuales = []
unidades = []
roles = []

for nombre in df.columns:
    if nombre in {"encounter_id", "patient_nbr"}:
        tipos_conceptuales.append("Identificador")
        unidades.append("Código")
        roles.append("Referencia de registro")
    elif nombre == "readmitted":
        tipos_conceptuales.append("Categórica")
        unidades.append("NO, <30, >30")
        roles.append("Variable de referencia")
    elif info_uci.loc[nombre, "type"] == "Integer":
        tipos_conceptuales.append("Cuantitativa")
        unidades.append("Días" if nombre == "time_in_hospital" else "Conteo")
        roles.append("Variable descriptiva")
    else:
        tipos_conceptuales.append("Categórica")
        unidades.append("Intervalos de 10 años" if nombre == "age" else "Categorías/códigos; ver significado")
        roles.append("Variable descriptiva")

diccionario = pd.DataFrame({
    "nombre": df.columns,
    "significado": info_uci.description.to_numpy(),
    "tipo_observado": df.dtypes.astype(str).to_numpy(),
    "tipo_conceptual": tipos_conceptuales,
    "unidades_o_categorías": unidades,
    "rol_analítico": roles,
})
display(diccionario)
assert len(diccionario) == df.shape[1] == 50
```

Los 101 766 registros son **hospitalizaciones** y corresponden a 71 518 identificadores de paciente distintos. El diccionario contiene las 50 columnas. Aunque algunos códigos de ingreso y alta se almacenen como enteros, conceptualmente identifican categorías. `age` aparece en intervalos y tampoco representa una edad continua directamente escalable.

```python
nulos = df.isna().sum()
preguntas = df.eq("?").sum()
calidad = pd.DataFrame({
    "columna": df.columns,
    "nulos": nulos.to_numpy(),
    "marcador_?": preguntas.to_numpy(),
})
calidad["total_ausente_observado"] = calidad.nulos + calidad["marcador_?"]
calidad["porcentaje"] = (100 * calidad.total_ausente_observado / len(df)).round(2)
display(calidad)

duplicados = pd.Series({
    "filas_completas": int(df.duplicated().sum()),
    "encounter_id": int(df.encounter_id.duplicated().sum()),
    "patient_nbr": int(df.patient_nbr.duplicated().sum())
}, name="repeticiones")
display(duplicados)
```

Los ausentes observados se concentran en `weight` (98 569; 96,86 %), `medical_specialty` (49 949; 49,08 %) y `payer_code` (40 256; 39,56 %). También hay ausencias en `race` (2 273), `diag_1` (21), `diag_2` (358) y `diag_3` (1 423). No quedan `?` literales en el DataFrame cargado.

La carga presenta nulos en `max_glu_serum` (96 420) y `A1Cresult` (84 748). Según el diccionario de UCI, el valor `None` en ambas variables expresa que la prueba **no se realizó**. Se recuperará esa categoría en la copia de trabajo; no se inventarán resultados de laboratorio.

Hay cero filas completas duplicadas y cero `encounter_id` repetidos. Los 30 248 registros cuyo `patient_nbr` ya apareció son encuentros adicionales, no 30 248 pacientes distintos repetidos. Se conservan las hospitalizaciones.

```python
estadisticas = pd.DataFrame({
    "mínimo": df[cuantitativas].min(),
    "Q1": df[cuantitativas].quantile(.25),
    "mediana": df[cuantitativas].median(),
    "media": df[cuantitativas].mean(),
    "Q3": df[cuantitativas].quantile(.75),
    "máximo": df[cuantitativas].max(),
    "desviación": df[cuantitativas].std(),
})
display(estadisticas.round(2))

frecuencia = df.readmitted.value_counts().reindex(["NO", ">30", "<30"])
display(pd.DataFrame({
    "hospitalizaciones": frecuencia,
    "porcentaje": (100 * frecuencia / len(df)).round(2)
}))
```

La mediana de procedimientos de laboratorio es 44, frente a una media de 43,10, y el máximo observado es 132. Para medicamentos, la mediana es 15, la media 16,02 y el máximo 81. Estas escalas y extremos merecen una revisión antes de transformar.

Las categorías `NO`, `>30` y `<30` representan respectivamente 54 864 (53,91 %), 35 545 (34,93 %) y 11 357 (11,16 %) hospitalizaciones. La frecuencia de una categoría no explica por sí sola las causas de la readmisión.

```python
variables_graficas = ["num_lab_procedures", "num_medications"]
fig, ejes = plt.subplots(2, 2, figsize=(12, 7))

for fila, variable in enumerate(variables_graficas):
    sns.histplot(df[variable], bins=30, ax=ejes[fila, 0], color="#146c8a")
    ejes[fila, 0].set(title=f"Histograma · {variable}", xlabel=variable, ylabel="Hospitalizaciones")
    sns.boxplot(x=df[variable], ax=ejes[fila, 1], color="#a9cfdb")
    ejes[fila, 1].set(title=f"Caja · {variable}", xlabel=variable)

plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(7, 4))
sns.countplot(data=df, x="readmitted", order=["NO", ">30", "<30"], ax=ax, color="#146c8a")
ax.set(title="Categorías de readmisión", xlabel="Readmisión", ylabel="Hospitalizaciones")
plt.tight_layout()
plt.show()
```

**Interpretación de las cinco visualizaciones.** El histograma de procedimientos de laboratorio concentra la mayor parte de los casos alrededor de sus valores centrales y muestra registros hasta 132; su diagrama de caja destaca los que exceden el rango habitual. El histograma de medicamentos concentra muchos registros alrededor de 10 a 20 y tiene una cola hacia cantidades altas; su caja representa la mitad central entre 10 y 20, con mediana 15 y puntos altos separados. El gráfico de `readmitted` confirma que `NO` tiene más hospitalizaciones que `>30`, y que `<30` es la menos frecuente. Los puntos apartados requieren interpretación, no eliminación automática.

```python
recuento_iqr = []
for variable in variables_graficas:
    q1, q3 = df[variable].quantile([.25, .75])
    ancho = q3 - q1
    minimo_iqr, maximo_iqr = q1 - 1.5 * ancho, q3 + 1.5 * ancho
    alejados = ~df[variable].between(minimo_iqr, maximo_iqr)
    recuento_iqr.append({
        "variable": variable, "Q1": q1, "Q3": q3, "IQR": ancho,
        "límite_inferior": minimo_iqr, "límite_superior": maximo_iqr,
        "cantidad": int(alejados.sum()),
        "porcentaje": 100 * alejados.mean()
    })
display(pd.DataFrame(recuento_iqr).round(2))
```

El método IQR señala **143 registros (0,14 %)** con más de 96 procedimientos de laboratorio y **2 557 (2,51 %)** con más de 35 medicamentos. Los límites inferiores son negativos y no afectan a estos conteos no negativos. Estas observaciones se conservarán: un umbral estadístico no acredita un error clínico o de captura. Su posible influencia sobre Min-Max y Z-score se documenta.

```python
por_readmision = df.groupby("readmitted")[["num_lab_procedures", "num_medications"]].agg(["mean", "median"])
display(por_readmision.round(2))
```

La categoría `<30` tiene medias de **44,23** procedimientos de laboratorio y **16,90** medicamentos por hospitalización. En `NO` las medias son **42,38** y **15,67**. Las medianas son 45 y 16 en `<30`, frente a 44 y 14 en `NO`. `>30` presenta valores intermedios: medias 43,84 y 16,28, con medianas 45 y 15. Se trata de diferencias descriptivas entre hospitalizaciones; no permiten afirmar causalidad.

## Fase 3. Selección y limpieza

**Reglas:** conservar las hospitalizaciones distintas de un paciente, registrar las ausencias categóricas y mantener los puntos señalados por IQR. Recuperar `None` en las dos pruebas como categoría documentada. No imputar las siete variables cuantitativas si la revisión confirma que están completas. Conservar los identificadores como referencia, fuera de los escaladores.

```python
analisis = df.copy()
print("Antes de la decisión:", analisis.shape)
print("Nulos de las pruebas antes:")
display(analisis[["max_glu_serum", "A1Cresult"]].isna().sum())

for prueba in ["max_glu_serum", "A1Cresult"]:
    analisis[prueba] = analisis[prueba].fillna("None")

print("Después de recuperar la categoría:", analisis.shape)
print("Nulos de las pruebas después:")
display(analisis[["max_glu_serum", "A1Cresult"]].isna().sum())
print("Nulos en las siete variables:", int(analisis[cuantitativas].isna().sum().sum()))
print("Valores negativos:", int(analisis[cuantitativas].lt(0).sum().sum()))
print("Estancias fuera de 1–14 días:", int((~analisis.time_in_hospital.between(1, 14)).sum()))
print("Filas duplicadas después:", int(analisis.duplicated().sum()))

assert analisis.shape == df.shape == (101766, 50)
```

La copia de trabajo permanece en **101 766 filas y 50 columnas**. Se restituyeron `None` en 96 420 registros de `max_glu_serum` y en 84 748 de `A1Cresult`, respetando que indican prueba no realizada. Las otras ausencias se mantienen documentadas. Las siete variables de interés tienen cero nulos y cero valores negativos; la duración hospitalaria está en el intervalo documentado. No se retiraron columnas ni hospitalizaciones en esta decisión. La selección posterior produce siete columnas sin reducir las filas.

```python
ids_de_referencia = analisis[["encounter_id", "patient_nbr"]].copy()
numericas_limpias = analisis.loc[:, cuantitativas].copy()

print("Tabla de trabajo:", analisis.shape)
print("Variables para transformar:", numericas_limpias.shape)
display(numericas_limpias.head())

assert numericas_limpias.shape == (101766, 7)
assert numericas_limpias.index.equals(ids_de_referencia.index)
assert not numericas_limpias.isna().any().any()
```

`encounter_id` y `patient_nbr` identifican registros y pacientes; no son magnitudes. Tampoco se escalan `admission_type_id`, `discharge_disposition_id`, `admission_source_id` ni los intervalos de `age`. `numericas_limpias` conserva una copia anterior a ambas transformaciones.

```python
from sklearn.preprocessing import MinMaxScaler

escalador_01 = MinMaxScaler(feature_range=(0, 1))
normalizada = pd.DataFrame(
    escalador_01.fit_transform(numericas_limpias.copy()),
    index=numericas_limpias.index, columns=numericas_limpias.columns
)
display(normalizada.head(10).round(4))

limites_01 = pd.DataFrame({
    "mínimo_original": numericas_limpias.min(),
    "máximo_original": numericas_limpias.max(),
    "mínimo_01": normalizada.min(),
    "máximo_01": normalizada.max(),
})
display(limites_01.round(4))
assert normalizada.min().ge(-1e-9).all()
assert normalizada.max().le(1 + 1e-9).all()
assert normalizada.index.equals(numericas_limpias.index)
```

**Min-Max.** Para una columna no constante se usa `(x − mínimo) / (máximo − mínimo)`. Las siete columnas quedan entre 0 y 1 y conservan las filas y nombres. Los valores extremos fijan ambos extremos del intervalo, por lo que pueden comprimir los valores habituales. Si una columna fuera constante, la división manual tendría denominador cero; `MinMaxScaler` la gestiona y en este rango devuelve 0.

```python
from sklearn.preprocessing import StandardScaler

escalador_z = StandardScaler()
estandarizada = pd.DataFrame(
    escalador_z.fit_transform(numericas_limpias.copy()),
    index=numericas_limpias.index, columns=numericas_limpias.columns
)
display(estandarizada.head(10).round(4))

comprobacion_z = pd.DataFrame({
    "media": estandarizada.mean(),
    "desviación_ddof_0": estandarizada.std(ddof=0)
})
display(comprobacion_z.round(4))

constantes = numericas_limpias.nunique().eq(1)
assert estandarizada.mean().abs().lt(1e-9).all()
assert (estandarizada.loc[:, ~constantes].std(ddof=0) - 1).abs().lt(1e-9).all()
assert estandarizada.loc[:, constantes].std(ddof=0).abs().lt(1e-9).all()
assert estandarizada.index.equals(numericas_limpias.index)
```

**Z-score.** Se ajustó sobre `numericas_limpias`, independientemente de Min-Max. La fórmula para cada columna no constante es `(x − media) / desviación estándar`. Las medias resultan aproximadamente 0 y las desviaciones 1, calculadas con `ddof=0`. Son normales los valores negativos y los superiores a 1; la transformación no limita el rango. Una columna constante quedaría en 0, con desviación resultante 0. Los extremos pueden afectar media y desviación.

```python
versiones = {
    "Original limpio": numericas_limpias,
    "Min-Max": normalizada,
    "Z-score": estandarizada
}

resumen_escalas = pd.concat({
    nombre: pd.DataFrame({
        "mínimo": tabla.min(), "máximo": tabla.max(),
        "media": tabla.mean(), "desviación_ddof_0": tabla.std(ddof=0)
    })
    for nombre, tabla in versiones.items()
}, names=["versión", "variable"])
display(resumen_escalas.round(4))

fig, ejes = plt.subplots(2, 3, figsize=(15, 7))
for fila, variable in enumerate(variables_graficas):
    for col, (etiqueta, tabla) in enumerate(versiones.items()):
        ejes[fila, col].hist(tabla[variable], bins=30, color="#146c8a", edgecolor="white")
        ejes[fila, col].set(title=f"{variable} · {etiqueta}", xlabel="Valor", ylabel="Hospitalizaciones")
plt.tight_layout()
plt.show()
```

La comparación estadística conserva los rangos originales en la primera versión; Min-Max produce mínimos 0 y máximos 1, mientras Z-score genera medias próximas a 0 y desviaciones 1. En los histogramas de laboratorio y medicamentos se mantienen las concentraciones y colas; cambia la escala horizontal. Como ambas transformaciones son lineales por columna, no corrigen por sí mismas una distribución asimétrica.

```python
variable_ejemplo = "num_procedures"
valores = [0, 1, 2, 3, 4]
serie = numericas_limpias[variable_ejemplo]
minimo, maximo = serie.min(), serie.max()
media, sigma = serie.mean(), serie.std(ddof=0)

print(f"Parámetros de {variable_ejemplo}: mínimo={minimo}, máximo={maximo}, media={media:.10f}, sigma={sigma:.10f}")
filas_ejemplo = [serie.index[serie.eq(valor)][0] for valor in valores]
manual = pd.DataFrame({
    "índice_original": filas_ejemplo, "x": valores,
    "Min-Max fórmula": [(x - minimo)/(maximo - minimo) for x in valores],
    "Min-Max Python": normalizada.loc[filas_ejemplo, variable_ejemplo].to_numpy(),
    "Z-score fórmula": [(x - media)/sigma for x in valores],
    "Z-score Python": estandarizada.loc[filas_ejemplo, variable_ejemplo].to_numpy(),
})
display(manual.round(4))
assert np.allclose(manual["Min-Max fórmula"], manual["Min-Max Python"])
assert np.allclose(manual["Z-score fórmula"], manual["Z-score Python"])
```

**Demostración con cinco valores de `num_procedures`.** Mínimo = **0**, máximo = **6**, media = **1,3397303618** y desviación poblacional = **1,7057985981**.

| x | Sustitución Min-Max | Resultado | Sustitución Z-score | Resultado |
|---:|---|---:|---|---:|
| 0 | (0 − 0) / (6 − 0) | 0,0000 | (0 − 1,3397303618) / 1,7057985981 | −0,7854 |
| 1 | (1 − 0) / (6 − 0) | 0,1667 | (1 − 1,3397303618) / 1,7057985981 | −0,1992 |
| 2 | (2 − 0) / (6 − 0) | 0,3333 | (2 − 1,3397303618) / 1,7057985981 | 0,3871 |
| 3 | (3 − 0) / (6 − 0) | 0,5000 | (3 − 1,3397303618) / 1,7057985981 | 0,9733 |
| 4 | (4 − 0) / (6 − 0) | 0,6667 | (4 − 1,3397303618) / 1,7057985981 | 1,5595 |

Los cálculos directos y las salidas de los escaladores coinciden al redondear a cuatro decimales.

```python
normalizada.to_csv("datos_normalizados.csv", index=True, index_label="indice_original")
estandarizada.to_csv("datos_estandarizados.csv", index=True, index_label="indice_original")

archivo_01 = pd.read_csv("datos_normalizados.csv", index_col="indice_original")
archivo_z = pd.read_csv("datos_estandarizados.csv", index_col="indice_original")
print("Archivo Min-Max:", archivo_01.shape)
print("Archivo Z-score:", archivo_z.shape)

assert archivo_01.shape == archivo_z.shape == (len(df), 7)
assert archivo_01.index.equals(archivo_z.index)
assert archivo_01.index.equals(ids_de_referencia.index)
assert archivo_01.columns.tolist() == archivo_z.columns.tolist() == cuantitativas
print("Exportaciones verificadas: siete variables, filas y orden correspondientes.")
```

## Conclusiones y reflexión

Se distinguieron **101 766 hospitalizaciones** de **71 518 pacientes** y se revisaron los 50 campos. La categoría `NO` fue la más frecuente (53,91 %). En la comparación descriptiva, `<30` tuvo mayores medias de procedimientos de laboratorio y medicamentos que `NO`; esto no demuestra causalidad ni permite tomar decisiones clínicas.

El IQR señaló 143 registros en procedimientos de laboratorio y 2 557 en medicamentos. Se conservaron porque exceder un límite estadístico no demuestra error. Las siete variables requeridas estaban completas, por lo que no hubo imputación en ellas. Ambas exportaciones retuvieron 101 766 filas y el mismo índice original.

Min-Max es útil cuando se necesita el rango [0, 1]; Z-score representa distancias respecto de la media en unidades de desviación. Ninguno elimina atípicos o corrige asimetría. El periodo 1999–2008, los hospitales incluidos y la posible repetición de pacientes limitan la interpretación.

**Preguntas de reflexión.** (1) Un código categórico entero identifica una clase: su distancia numérica no mide una cantidad. (2) Un extremo fija el rango de Min-Max y afecta la media y desviación de Z-score. (3) El mismo paciente puede tener encuentros distintos sin duplicación de filas. (4) Reescalar cambia unidades, mientras corregir asimetría requiere otra transformación que esta práctica no pide. (5) Si se entrenase un modelo posteriormente, se separarían los datos antes de ajustar los escaladores, se transformarían validación y test con esos parámetros y se evitaría compartir pacientes entre particiones.

**Fuente:** Clore, J., Cios, K., DeShazo, J. y Strack, B. (2014). *Diabetes 130-US Hospitals for Years 1999–2008*. UCI Machine Learning Repository. https://doi.org/10.24432/C5230J
