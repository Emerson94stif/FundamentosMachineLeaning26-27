```python
# Cargar y explorar los datos
datos = datasets.load_diabetes(as_frame=True)
#pprint(datos)
#print(np.shape(datos.data))
df=datos.frame
display(df.head())
print("dimensiones", df.shape)
```

## Carga y revisión inicial

### Carga y exploración inicial del dataset

`datasets.load_diabetes(as_frame=True)` carga los datos de diabetes en un formato compatible con pandas. `datos.frame` reúne las diez variables predictoras y la columna objetivo `target` en una sola tabla llamada `df`.

`display(df.head())` muestra los cinco primeros registros para observar la estructura de los datos. `df.shape` muestra el tamaño de la tabla: 442 filas y 11 columnas.

Las líneas que comienzan con `#` están comentadas; por ello, `pprint(datos)` y `print(np.shape(datos.data))` no se ejecutan.

```python
# Extraer la matriz de datos "X" y la variable target "y"
X = datos.data
y = datos.target
```

## Selección de entradas y resultado esperado

Se organizan los datos en dos componentes.

- `X = datos.data` contiene las diez características utilizadas para elaborar las estimaciones.
- `y = datos.target` contiene el valor de progresión asociado a cada registro.

Durante el entrenamiento, el algoritmo aprende una relación entre las entradas de `X` y los valores de `y`.

```python
# Partición EXTERNA: hold-out split 80% train - 20% test.
X_train, x_test, y_train, y_test = train_test_split(X,y, test_size= 0.20, random_state=42)
print ("datos de entrenamioento:" , X_train.shape)
print ("datos de test:" , x_test.shape)
```

### División de los datos

`train_test_split(X, y, test_size=0.20, random_state=42)` separa los registros en dos grupos: el 80 % para entrenamiento y el 20 % para prueba. `random_state=42` permite reproducir la misma división al ejecutar nuevamente el notebook.

El resultado es de 353 registros para entrenamiento y 89 para prueba. El conjunto de prueba se reserva para evaluar el modelo después de entrenarlo.

```python
# Estandarización de los datos de entrenamiento y test
standarizer = preprocessing.StandardScaler()
```

## Configuración del escalado

Se crea un objeto `StandardScaler` para estandarizar las variables predictoras.

El escalador calcula la media y la desviación estándar de cada característica durante su ajuste y utiliza esos valores para transformarla.

En esta celda se prepara el objeto. Su ajuste y aplicación se realizarán dentro del pipeline durante el entrenamiento.

```python
# Métricas de evaluación.
metricas = {
  'MAE': 'neg_mean_absolute_error',
  'RMSE': make_scorer(lambda y, y_pred:
                      sqrt(mean_squared_error(y, y_pred)),
                      greater_is_better=False),
  'MAPE': make_scorer(lambda y, y_pred:
                      np.mean(np.abs((y - y_pred) / y)) * 100,
                      greater_is_better=False)}
```

### Definición de métricas de evaluación

El diccionario `metricas` establece tres formas de medir la diferencia entre los valores reales y las predicciones:

- **MAE:** promedio de las diferencias absolutas.
- **RMSE:** raíz del promedio de los errores al cuadrado; da más peso a errores grandes.
- **MAPE:** error absoluto expresado como porcentaje respecto al valor real.

`make_scorer` permite utilizar las fórmulas personalizadas de RMSE y MAPE en la validación cruzada. Se configura `greater_is_better=False` porque un error menor representa un mejor resultado. Por la convención de scikit-learn, los valores que devuelve `cross_validate` para estas métricas aparecen con signo negativo; para interpretarlos como errores se considera su magnitud.

```python
# Definimos un algoritmo SVM de regresión
alg =Pipeline([
    ("standarizer", standarizer),
    ("regresion", SVR(C=10, gamma= "scale", kernel="rbf"))
    ])
```

### Construcción del modelo de regresión

`Pipeline` organiza dos pasos consecutivos. Primero, `StandardScaler` estandariza las variables. Después, `SVR` aprende a estimar la variable objetivo.

`SVR` es una máquina de vectores de soporte para **regresión**. En esta configuración utiliza `C=10`, `gamma="scale"` y un kernel `rbf`. El pipeline ayuda a que la estandarización se ajuste con los datos de entrenamiento correspondientes en cada partición de la validación.

```python
# Validación y evaluación del modelo.
# en "cv = KFold(n_splits=5)" se hace un cross-validation INTERNO!! 
results = cross_validate(alg, X_train, y_train, cv = KFold(n_splits=5, shuffle=True, random_state=42), scoring = metricas)
```

```python
# Presentación de los resultados de la evaluación.
pprint(results)
```

```python
# Una vez hemos optimizado los hiperparámetros en el conjunto de validación, entrenamos un nuevo modelo que contenga
# todos los datos de entrenamiento (es decir, train + validation) para evaluar el test

#model = SVR(C=10, gamma='scale', kernel='rbf')
model= clone(alg)
model = model.fit(X_train, y_train)
```

### Entrenamiento con todos los datos disponibles para aprender

`clone(alg)` crea una nueva instancia del pipeline con la misma configuración. Luego, `fit(X_train, y_train)` la entrena utilizando los 353 registros del conjunto de entrenamiento.

La celda prepara el modelo que se evaluará con el conjunto de prueba. Aunque el comentario del código menciona una optimización de hiperparámetros, en el notebook no se muestra una búsqueda de valores para `C`, `gamma` o `kernel`; se conserva la configuración establecida anteriormente.

```python
# Extraemos el error absoluto promedio (MAE) del conjunto de test
y_pred = model.predict(x_test)
mae = mean_absolute_error(y_test,y_pred)

print('Test results (MAE): ', mae)
print("test results (RMSE):",round(sqrt(mean_squared_error(y_test, y_pred)),3))
#print("Test results (MAP%):",round (mean_absolute_percentage_error(y_test, y_pred) *100, 3))
```

### Predicción y evaluación final

`model.predict(x_test)` genera una predicción para cada uno de los 89 registros reservados para prueba. Después, `mean_absolute_error` calcula el MAE y `mean_squared_error`, junto con `sqrt`, permite calcular el RMSE.

En la ejecución guardada en el notebook se obtuvo un **MAE aproximado de 41,32** y un **RMSE aproximado de 51,79**, medidos en la escala numérica de `target`. Son medidas del error de predicción de progresión; no indican un porcentaje de acierto al diagnosticar diabetes.

```python
# También podemos extraer las predicciones para cuánto difieren los valores predichos de los reales
print('Predicciones:   ', np.round(y_pred[:10], 1))
print('Valores reales: ', y_test[:10])
```

### Comparación de las primeras predicciones

`np.round(y_pred[:10], 1)` selecciona las primeras diez predicciones y las muestra con un decimal. `y_test[:10]` presenta los diez valores reales correspondientes.

Esta comparación permite observar, caso por caso, cuánto se aproxima el modelo al valor registrado. Son ejemplos individuales y no sustituyen las métricas calculadas sobre todo el conjunto de prueba.

```python
# Grafica de realidad -vs- prediccion.
def grafica_real_vs_pred(y_true, y_pred, metricas, algoritmo):
    fig, ax = plt.subplots()
    ax.scatter(y_true, y_pred, edgecolors=(0, 0, 0))
    ax.plot([y_true.min(), y_true.max()], [y_true.min(), y_true.max()], 'k--', lw=4)
    ax.set_xlabel('Valor real de la clase')
    ax.set_ylabel('Predicción')
    plt.show()
```

```python
grafica_real_vs_pred(y_test, y_pred, metricas, model)
```

## Representación del desempeño

La llamada `grafica_real_vs_pred(y_test, y_pred, metricas, model)` utiliza los resultados del conjunto de prueba para generar el gráfico.

La cercanía de los puntos a la diagonal permite apreciar el ajuste de las estimaciones. Los puntos alejados representan errores más grandes.

Los puntos por encima de la diagonal corresponden a sobreestimaciones. Los situados por debajo corresponden a subestimaciones.
