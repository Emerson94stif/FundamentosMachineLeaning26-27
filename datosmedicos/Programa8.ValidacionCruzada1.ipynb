**Validacion Cruzada:** Es una técnica de evaluación de modelos que divide los datos en múltiples particiones (folds) para entrenar y validar repetidamente, logrando una estimación más estable y generalizable del rendimiento del modelo.

**Formula:**

```python
from sklearn.model_selection import KFold
```

```python
# Carga de datos.
X = ["a", "b", "c", "d", "e", "f", "g", "h", "i", "j"]
```

### Validación cruzada con K-Fold

Se configura `KFold` para dividir los registros de `X` en cinco partes o *bolsas*. Con `shuffle=True`, los registros se mezclan antes de formar las divisiones.

`kf.split(X)` crea un generador que entregará cinco pares de índices. En cada vuelta, cuatro bolsas corresponden al entrenamiento y una a la validación; la bolsa de validación cambia hasta que todas hayan ocupado ese lugar una vez.

`print(bolsas)` muestra la representación del generador, no los índices de las cinco divisiones. Esta celda solo prepara las particiones: todavía no entrena ni evalúa un modelo. Como no se fijó `random_state`, las particiones pueden variar entre ejecuciones.

```python
# Validación cruzada.
kf = KFold(n_splits = 5, shuffle=True) # Shuffle permite aleatorizar las bolsas de test vez de hacer una lectura por orden 
bolsas = kf.split(X)
print(bolsas)
```

```python
# Mostrar ejemplos de cada iteración.
k = 1
for train, test in bolsas:
    print("Iteracion", k, ":")
    print(" - Entrenamiento: %s" % (train)) 
    print(" - Test: %s" % (test))
    k = k + 1
```
